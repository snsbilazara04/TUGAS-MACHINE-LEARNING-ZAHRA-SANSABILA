```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Membaca dataset
train_input = pd.read_csv("train_input.csv")
train_label = pd.read_csv("train_label.csv")
test_input = pd.read_csv("test_input.csv")
order = pd.read_csv("order.csv")

# Menampilkan 5 data pertama
print("TRAIN INPUT")
display(train_input.head())

print("TRAIN LABEL")
display(train_label.head())

print("TEST INPUT")
display(test_input.head())

print("ORDER")
display(order.head())
```

```python
print("Ukuran train_input:", train_input.shape)
print("Ukuran train_label:", train_label.shape)
print("Ukuran test_input:", test_input.shape)
print("Ukuran order:", order.shape)
```

```python
# Menampilkan ringkasan informasi dataset
train_input.info()
```

```python
# Menampilkan statistik deskriptif
train_input.describe(include="all")
```

```python
# Menghitung jumlah missing value
missing_values = train_input.isnull().sum()

print("Jumlah missing value pada setiap kolom:")
print(missing_values)

print("\nKolom yang memiliki missing value:")
print(missing_values[missing_values > 0])
```

```python
# Mengidentifikasi baris duplikat
duplicates = train_input.duplicated()

print("Jumlah baris duplikat:", duplicates.sum())

# Menampilkan data duplikat jika ada
train_input[duplicates]
```

```python
# Menghapus data duplikat
train_input = train_input.drop_duplicates()

print("Jumlah data setelah menghapus duplikat:", len(train_input))
```

```python
# Kolom numerik yang digunakan sebagai fitur
numeric_features = [
    'Hours Studied',
    'Previous Scores',
    'Sleep Hours',
    'Sample Question Papers Practiced'
]

print("Kolom numerik:")
print(numeric_features)
```

```python
# Menentukan kolom kategorikal
category_features = [
    'Extracurricular Activities'
]

print("Kolom kategorikal:")
print(category_features)
```

```python
# Box plot setiap fitur numerik

for feature in numeric_features:
    plt.figure(figsize=(10, 5))
    
    sns.boxplot(x=train_input[feature])
    
    plt.title(f'Box Plot of {feature}')
    plt.xlabel(feature)
    plt.show()
```

```python
# Menghitung Q1 dan Q3
Q1 = train_input[numeric_features].quantile(0.25)
Q3 = train_input[numeric_features].quantile(0.75)

# Menghitung IQR
IQR = Q3 - Q1

# Menentukan batas bawah dan batas atas
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Menghitung jumlah outlier
outliers = (
    (train_input[numeric_features] < lower_bound) |
    (train_input[numeric_features] > upper_bound)
).sum()

print("Jumlah outlier pada setiap kolom:")
print(outliers)
```

```python
# Membuat kondisi data tanpa outlier
condition = ~(
    (train_input[numeric_features] < lower_bound) |
    (train_input[numeric_features] > upper_bound)
).any(axis=1)

# Menghapus data yang mengandung outlier
train_input_clean = train_input.loc[condition].copy()

print("Jumlah data sebelum menghapus outlier:", len(train_input))
print("Jumlah data setelah menghapus outlier:", len(train_input_clean))
```

```python
# Membuat salinan dataset yang sudah dibersihkan
df_eda = train_input_clean.copy()

print("Ukuran dataset setelah cleaning:", df_eda.shape)
```

```python
# Menghitung jumlah variabel numerik
num_vars = len(numeric_features)

# Menentukan jumlah kolom subplot
n_cols = 2

# Menghitung jumlah baris
n_rows = -(-num_vars // n_cols)

# Membuat subplot
fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(14, n_rows * 4)
)

# Mengubah axes menjadi array 1 dimensi
axes = np.array(axes).flatten()

# Membuat histogram
for i, column in enumerate(numeric_features):
    df_eda[column].hist(
        ax=axes[i],
        bins=20,
        edgecolor='black'
    )
    
    axes[i].set_title(column)
    axes[i].set_xlabel('Value')
    axes[i].set_ylabel('Frequency')

# Menghapus subplot yang tidak digunakan
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
# Visualisasi distribusi fitur numerik

plt.figure(figsize=(15, 8))

for i, column in enumerate(numeric_features, 1):
    plt.subplot(2, 2, i)
    
    sns.histplot(
        df_eda[column],
        kde=True,
        bins=30
    )
    
    plt.title(f'Distribution of {column}')
    plt.xlabel(column)
    plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
# Distribusi Performance Index

plt.figure(figsize=(10, 5))

sns.histplot(
    df_eda['Performance Index'],
    kde=True,
    bins=30
)

plt.title('Distribution of Performance Index')
plt.xlabel('Performance Index')
plt.ylabel('Frequency')

plt.show()
```

```python
# Melakukan One-Hot Encoding
df_one_hot = pd.get_dummies(
    df_eda,
    columns=category_features,
    drop_first=False
)

# Menampilkan hasil
display(df_one_hot.head())

print("Ukuran data setelah One-Hot Encoding:", df_one_hot.shape)
```

```python
from sklearn.preprocessing import LabelEncoder

# Membuat salinan dataset
df_lencoder = df_eda.copy()

# Membuat Label Encoder
label_encoder = LabelEncoder()

# Melakukan Label Encoding
for col in category_features:
    df_lencoder[col] = label_encoder.fit_transform(
        df_lencoder[col]
    )

# Menampilkan hasil
display(df_lencoder.head())
```

```python
# Menghitung missing value
missing_values = df_lencoder.isnull().sum()

# Menghitung persentase missing value
missing_percentage = (
    missing_values / len(df_lencoder)
) * 100

# Membuat tabel missing value
missing_data = pd.DataFrame({
    'Missing Values': missing_values,
    'Percentage': missing_percentage
}).sort_values(
    by='Missing Values',
    ascending=False
)

# Menampilkan kolom yang memiliki missing value
display(
    missing_data[
        missing_data['Missing Values'] > 0
    ]
)
```

```python
# Menentukan jumlah variabel
num_vars = df_lencoder.shape[1]

# Jumlah kolom
n_cols = 3

# Jumlah baris
n_rows = -(-num_vars // n_cols)

# Membuat subplot
fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(18, n_rows * 4)
)

axes = np.array(axes).flatten()

# Membuat histogram
for i, column in enumerate(df_lencoder.columns):
    df_lencoder[column].hist(
        ax=axes[i],
        bins=20,
        edgecolor='black'
    )
    
    axes[i].set_title(column)
    axes[i].set_xlabel('Value')
    axes[i].set_ylabel('Frequency')

# Menghapus subplot kosong
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
# Mengambil kolom numerik
numeric_data = df_lencoder.select_dtypes(
    include='number'
)

# Menghitung korelasi
correlation_matrix = numeric_data.corr()

# Membuat heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    vmin=-1,
    vmax=1
)

plt.title('Correlation Matrix')
plt.show()
```

```python
# Menghitung korelasi terhadap target
target_corr = df_lencoder.corr()['Performance Index']

# Mengurutkan berdasarkan nilai absolut
target_corr_sorted = (
    target_corr.abs()
    .sort_values(ascending=False)
)

# Membuat grafik
plt.figure(figsize=(10, 6))

target_corr_sorted.plot(
    kind='bar'
)

plt.title('Correlation with Performance Index')
plt.xlabel('Variables')
plt.ylabel('Correlation Coefficient')

plt.tight_layout()
plt.show()
```

```python
# Memisahkan fitur (X) dan target (y)

X = df_lencoder.drop(
    columns=['ID', 'Performance Index']
)

y = df_lencoder['Performance Index']

print("Jumlah data X:", X.shape)
print("Jumlah data y:", y.shape)

print("\nKolom X:")
print(X.columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

# Membagi dataset menjadi data training dan testing
x_train, x_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=1
)

print("Jumlah data training:", len(x_train))
print("Jumlah data testing:", len(x_test))
```

```python
# Menghitung jumlah data keseluruhan
print("Jumlah data:", len(X))

# Menghitung jumlah data latih
print("Jumlah data latih:", len(x_train))

# Menghitung jumlah data test
print("Jumlah data test:", len(x_test))
```

```python
from sklearn.preprocessing import StandardScaler

# Membuat StandardScaler
scaler = StandardScaler()

# Melatih scaler hanya menggunakan data training
x_train_scaled = scaler.fit_transform(x_train)

# Menggunakan scaler yang sama pada data testing
x_test_scaled = scaler.transform(x_test)

# Mengubah kembali menjadi DataFrame
x_train_scaled = pd.DataFrame(
    x_train_scaled,
    columns=x_train.columns,
    index=x_train.index
)

x_test_scaled = pd.DataFrame(
    x_test_scaled,
    columns=x_test.columns,
    index=x_test.index
)

print("Standardisasi berhasil dilakukan.")

display(x_train_scaled.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Menentukan jumlah kolom dan baris secara otomatis
num_vars = len(x_train_scaled.columns)

n_cols = 3
n_rows = -(-num_vars // n_cols)

# Membuat figure
fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(18, n_rows * 5)
)

# Mengubah axes menjadi array 1 dimensi
axes = np.array(axes).flatten()

# Membuat histogram setiap fitur
for i, column in enumerate(x_train_scaled.columns):
    sns.histplot(
        x_train_scaled[column],
        kde=True,
        bins=30,
        ax=axes[i]
    )

    axes[i].set_title(f'Standardized {column}')
    axes[i].set_xlabel(column)
    axes[i].set_ylabel('Frequency')

# Menghapus kotak grafik yang tidak digunakan
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
from sklearn import linear_model

# Membuat model LARS
lars = linear_model.Lars(
    n_nonzero_coefs=1
)

# Melatih model
lars.fit(
    x_train_scaled,
    y_train
)

print("Model LARS berhasil dilatih.")
```

```python
from sklearn.linear_model import LinearRegression

# Membuat model Linear Regression
LR = LinearRegression()

# Melatih model
LR.fit(
    x_train_scaled,
    y_train
)

print("Model Linear Regression berhasil dilatih.")
```

```python
from sklearn.ensemble import GradientBoostingRegressor

# Membuat model Gradient Boosting Regressor
GBR = GradientBoostingRegressor(
    random_state=184
)

# Melatih model
GBR.fit(
    x_train_scaled,
    y_train
)

print("Model Gradient Boosting Regressor berhasil dilatih.")
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Prediksi menggunakan LARS
pred_lars = lars.predict(x_test_scaled)

# Menghitung MAE
mae_lars = mean_absolute_error(
    y_test,
    pred_lars
)

# Menghitung MSE
mse_lars = mean_squared_error(
    y_test,
    pred_lars
)

# Menghitung R2
r2_lars = r2_score(
    y_test,
    pred_lars
)

# Membuat DataFrame hasil evaluasi
data = {
    'MAE': [mae_lars],
    'MSE': [mse_lars],
    'R2': [r2_lars]
}

df_results = pd.DataFrame(
    data,
    index=['LARS']
)

display(df_results)
```

```python
# Prediksi menggunakan Linear Regression
pred_LR = LR.predict(x_test_scaled)

# Menghitung MAE
mae_LR = mean_absolute_error(
    y_test,
    pred_LR
)

# Menghitung MSE
mse_LR = mean_squared_error(
    y_test,
    pred_LR
)

# Menghitung R2
r2_LR = r2_score(
    y_test,
    pred_LR
)

# Menambahkan hasil ke DataFrame
df_results.loc[
    'Linear Regression'
] = [
    mae_LR,
    mse_LR,
    r2_LR
]

display(df_results)
```

```python
# Prediksi menggunakan Gradient Boosting Regressor
pred_GBR = GBR.predict(x_test_scaled)

# Menghitung MAE
mae_GBR = mean_absolute_error(
    y_test,
    pred_GBR
)

# Menghitung MSE
mse_GBR = mean_squared_error(
    y_test,
    pred_GBR
)

# Menghitung R2
r2_GBR = r2_score(
    y_test,
    pred_GBR
)

# Menambahkan hasil ke DataFrame
df_results.loc[
    'Gradient Boosting Regressor'
] = [
    mae_GBR,
    mse_GBR,
    r2_GBR
]

display(df_results)
```

```python
# Membulatkan hasil evaluasi
df_results_rounded = df_results.round(4)

display(df_results_rounded)
```

```python
# Model dengan MAE terkecil
best_mae = df_results['MAE'].idxmin()

# Model dengan MSE terkecil
best_mse = df_results['MSE'].idxmin()

# Model dengan R2 terbesar
best_r2 = df_results['R2'].idxmax()

print("Model terbaik berdasarkan MAE :", best_mae)
print("Model terbaik berdasarkan MSE :", best_mse)
print("Model terbaik berdasarkan R2  :", best_r2)
```

```python
# Membuat salinan data test
test_data = test_input.copy()

# Mengambil ID
test_ids = test_data['ID']

# Menghapus ID karena tidak digunakan sebagai fitur
test_features = test_data.drop(
    columns=['ID']
)

# Label Encoding kolom kategorikal
for col in category_features:
    test_features[col] = label_encoder.transform(
        test_features[col]
    )

# Standardisasi menggunakan scaler yang sudah dilatih
test_features_scaled = scaler.transform(
    test_features
)

# Prediksi menggunakan GBR
test_predictions = GBR.predict(
    test_features_scaled
)

# Membuat DataFrame hasil prediksi
prediction_result = pd.DataFrame({
    'ID': test_ids,
    'Performance Index': test_predictions
})

display(prediction_result.head())
```

```python
prediction_result['Performance Index'] = (
    prediction_result['Performance Index']
    .round()
    .astype(int)
)

display(prediction_result.head(10))
```
